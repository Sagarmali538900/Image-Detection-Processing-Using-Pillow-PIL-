```python
from PIL import Image
import os
import hashlib
import shutil

def image_hash(image_path, hash_size=8):
    """
    Calculate the hash of an image.

    Args:
    image_path: Path to the image file.
    hash_size: Size of the hash in bits.

    Returns:
    A hash value as a hexadecimal string.
    """
    image = Image.open(image_path)
    image = image.resize((hash_size, hash_size), Image.ANTIALIAS)
    image = image.convert("L")
    pixels = list(image.getdata())
    avg_pixel = sum(pixels) / len(pixels)
    bits = "".join(['1' if pixel > avg_pixel else '0' for pixel in pixels])
    hexadecimal = int(bits, 2).__format__('016x').upper()
    return hexadecimal

def find_duplicates(source_folder, duplicate_folder):
    """
    Find duplicate images in a folder and move them to another folder.

    Args:
    source_folder: Path to the source folder containing images.
    duplicate_folder: Path to the folder where duplicate images will be moved.
    """
    hash_dict = {}
    for filename in os.listdir(source_folder):
        if filename.endswith(('.jpg', '.jpeg', '.png','.JPG','.PNG','.JPEG')):
            filepath = os.path.join(source_folder, filename)
            image_hash_value = image_hash(filepath)
            image_resolution = Image.open(filepath).size
            if image_hash_value in hash_dict:
                existing_filepath, existing_resolution = hash_dict[image_hash_value]
                if image_resolution[0] * image_resolution[1] > existing_resolution[0] * existing_resolution[1]:
                    shutil.move(existing_filepath, duplicate_folder)
                    hash_dict[image_hash_value] = (filepath, image_resolution)
                    print(f"Moved duplicate image: {os.path.basename(existing_filepath)}")
                else:
                    shutil.move(filepath, duplicate_folder)
                    print(f"Moved duplicate image: {filename}")
            else:
                hash_dict[image_hash_value] = (filepath, image_resolution)

# Set your source and duplicate folder paths
source_folder = "D:\Sukha_Mukt_Rajasthan\akoda\Canon"
duplicate_folder = "D:\Sukha_Mukt_Rajasthan\akoda\Canon\Duplicates"

# Ensure the duplicate folder exists
os.makedirs(duplicate_folder, exist_ok=True)

# Find and move duplicate images
print("Finding and moving duplicate images...")
find_duplicates(source_folder, duplicate_folder)
print("Duplicate images detection and moving completed.")
```

```python
def compare_folders(source_folder1, source_folder2, duplicate_folder):
    """
    Compare images in two folders and move duplicates to another folder.

    Args:
    source_folder1: Path to the first source folder containing images.
    source_folder2: Path to the second source folder containing images.
    duplicate_folder: Path to the folder where duplicate images will be moved.
    """
    hash_dict = {}

    # Function to process images in a folder
    def process_folder(folder):
        for filename in os.listdir(folder):
            if filename.endswith(('.jpg', '.jpeg', '.png','.JPG','.PNG','.JPEG')):
                filepath = os.path.join(folder, filename)
                image_hash_value = image_hash(filepath)
                image_resolution = Image.open(filepath).size
                if image_hash_value in hash_dict:
                    existing_filepath, existing_resolution = hash_dict[image_hash_value]
                    if image_resolution[0] * image_resolution[1] > existing_resolution[0] * existing_resolution[1]:
                        shutil.move(existing_filepath, duplicate_folder)
                        hash_dict[image_hash_value] = (filepath, image_resolution)
                        print(f"Moved duplicate image: {os.path.basename(existing_filepath)}")
                    else:
                        shutil.move(filepath, duplicate_folder)
                        print(f"Moved duplicate image: {filename}")
                else:
                    hash_dict[image_hash_value] = (filepath, image_resolution)

    # Process images in the first folder
    process_folder(source_folder1)
    
    # Process images in the second folder
    process_folder(source_folder2)

# Set your source folders and duplicate folder paths
source_folder1 = "D:\CANON"
source_folder2 = "D:\Photos"
duplicate_folder = "D:\Photos\Duplicates"

# Ensure the duplicate folder exists
os.makedirs(duplicate_folder, exist_ok=True)

# Compare folders and find duplicates
print("Comparing folders and moving duplicate images...")
compare_folders(source_folder1, source_folder2, duplicate_folder)
print("Duplicate images detection and moving completed.")
```
